### Grafico de dispersion

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.info()
```

```python
plt.scatter(titanic['age'], titanic['fare'])
plt.xlabel('Edad')
plt.ylabel('Tarifa')
plt.title('Relacion entre edad y tarifa');
```

## Seaborn

```python
sns.boxplot(x='survived', y='age', data=titanic, palette='Set2')
plt.title('Distribución de Edad por Supervivencia')
plt.show()
```

### Barras de supervivencia por clase

```python
sns.barplot(x='pclass', y='survived', data=titanic, errorbar=None, hue='pclass', legend=True, palette='muted')
plt.title('Tasa de supervivencia por clase')
plt.show()
```

### Plotly - Graficos interactivos

```python
fig = px.scatter(titanic, x='age', y='fare', color='survived', title='Relación Edad-Tarifa por supervivencia', labels={'survived': 'Sobrevivió'})
fig.show()
```

```python
fig1 = px.scatter_3d(titanic, x='age', y='fare', z='pclass', color='survived', symbol='sex', title='Análisis 3D - Titanic')
fig1.show()
```
